## 1. Values, types, and names

A **variable** is a name attached to a value. `=` means *assign*, not *equals*: the right side is computed first, then the name on the left points to the result.

```python
price = 12.50
quantity = 40
firm = "Palouse Lentil Co."
competitive = True

revenue = price * quantity
print(revenue)
print(type(price), type(quantity), type(firm), type(competitive))
```

The four basic types are `float` (decimal numbers), `int` (whole numbers), `str` (text), and `bool` (`True`/`False`). Multiplying a float by an int gives a float.

Because the right side is computed first, a name can appear on both sides:

```python
quantity = quantity + 5   # take the old value, add 5, store the result under the same name
quantity += 5             # shorthand for the same thing
quantity
```

**Floats are approximations.** Computers store most decimals in binary, so tiny rounding errors appear. Never test two floats for exact equality; check whether they are within a tolerance. Every numerical solver in this course stops at a tolerance for the same reason.

```python
print(0.1 + 0.2 == 0.3)
print(0.1 + 0.2)
print(abs((0.1 + 0.2) - 0.3) < 1e-12)
```

### Lists and dictionaries

A **list** holds values in order; positions start at **0**. A **dictionary** stores values under names, which is a convenient way to keep a model's parameters together.

```python
prices = [2.0, 2.5, 3.0, 3.5]
print(prices[0], prices[-1], len(prices))   # first, last, how many

params = {"a": 100, "b": 2, "c": 10, "d": 1}
print(params["a"], params["d"])
```

### Printing readably

An **f-string** (a string starting with `f`) puts values inside text. The part after `:` controls the format: `.2f` means two decimals, `,` adds thousands separators.

```python
print(f"{firm} earns ${price * quantity:,.2f} selling {quantity} units.")
```

## 2. Making decisions: `if`, `elif`, `else`

A price ceiling matters only if it is set below the equilibrium price. Code makes that decision with `if`:

```python
p_star = 30      # equilibrium price from Week 1
ceiling = 25

if ceiling < p_star:
    print(f"The ceiling binds: price is held at {ceiling}.")
else:
    print(f"The ceiling does not bind: price stays at {p_star}.")
```

**Indentation is part of the code.** The indented lines belong to the `if` or the `else`. Positron indents four spaces for you after a colon.

| Comparison | Meaning | | Logic | Meaning |
|---|---|---|---|---|
| `==` | equal | | `and` | both true |
| `!=` | not equal | | `or` | at least one true |
| `<`, `<=`, `>`, `>=` | ordering | | `not` | flips true/false |

Use `elif` ("else if") for more than two cases. Here we classify a price elasticity of demand:

```python
e = -0.6

if abs(e) > 1:
    label = "elastic"
elif abs(e) < 1:
    label = "inelastic"
else:
    label = "unit elastic"

print(f"An elasticity of {e} is {label}.")
```

## 3. Repeating work: `for` loops

A `for` loop runs the indented block once for each item in a sequence. With linear demand $Q = a - bP$:

```python
a, b = 100, 2

for p in [10, 20, 30, 40]:
    q = a - b * p
    print(f"P = {p:>3}:  Q = {q:>4},  revenue = {p * q:>5}")
```

`range(start, stop)` produces whole numbers from `start` up to, **but not including**, `stop`. A common pattern is to keep track of the best value found so far. This loop finds the revenue-maximizing whole-number price:

```python
best_p, best_rev = None, -1

for p in range(0, 51):          # 0, 1, ..., 50
    rev = p * (a - b * p)
    if rev > best_rev:
        best_p, best_rev = p, rev

print(f"Revenue is highest at P = {best_p}, where it equals {best_rev}.")
```

Check it with calculus: revenue $R = P(a - bP)$ has $dR/dP = a - 2bP = 0$ at $P = a/(2b) = 25$. The loop and the derivative agree. When a model has no closed-form answer, the loop still works; that is the main reason to compute.

A **list comprehension** builds a list from a loop in one line:

```python
revenues = [p * (a - b * p) for p in range(0, 51)]
print(max(revenues), revenues.index(max(revenues)))   # largest value, and its position
```

## 4. Repeating until done: `while` loops

A `while` loop keeps going as long as its condition is true. Use it when you don't know in advance how many steps you need. How many years until savings double at 5% interest?

```python
balance, years = 1.0, 0
rate = 0.05

while balance < 2:
    balance = balance * (1 + rate)
    years += 1

print(f"{years} years (the rule of 70 says about {70 / (rate * 100):.0f})")
```

Make sure something inside the loop moves you toward ending it. If a cell runs forever, press the **Interrupt** (square stop) button at the top of the notebook, fix the loop, and rerun.

## 5. Functions

A **function** gives a calculation a name, takes inputs (**parameters**), and hands back a result with `return`. Once written and tested, it can be reused with any inputs.

```python
def revenue(p, a, b):
    """Total revenue at price p when demand is Q = a - b*p."""
    return p * (a - b * p)

print(revenue(25, 100, 2))
print(revenue(25, a=120, b=3))     # arguments can be passed by name
```

The text in triple quotes is a **docstring**. It says what the function does; hover over the function name in Positron to see it.

**`return` is not `print`.** A function that only prints shows a number but gives back nothing (`None`), so you can't use its result in later calculations:

```python
def revenue_printed(p, a, b):
    print(p * (a - b * p))

x = revenue_printed(25, 100, 2)
print("x is", x)
```

**Default values** let you leave out arguments that rarely change. This cost function has a fixed cost and a quadratic variable cost, $C(q) = F + cq^2$:

```python
def cost(q, fixed=200, c=0.5):
    """Total cost of producing q units."""
    return fixed + c * q**2

print(cost(10), cost(10, fixed=0), cost(10, c=1))
```

**Pass what a function needs as arguments.** Names created inside a function exist only inside it, and a function that quietly reads variables defined elsewhere in the notebook breaks when those change. That is why the PS1 functions take `a` and `b` as inputs.

## 6. Putting it together: solving an equation by bisection

A firm sells at price $p = 40$ with cost $C(q) = 200 + 0.5q^2$. Profit is $\pi(q) = pq - C(q)$. What is the smallest quantity at which the firm breaks even ($\pi(q) = 0$)?

**Bisection** needs only one fact: if profit is negative at `lo` and positive at `hi`, the break-even point lies between them. Check the midpoint, keep the half that still brackets the answer, and repeat until the bracket is tiny.

```python
def profit(q, p=40, fixed=200, c=0.5):
    """Profit at quantity q."""
    return p * q - cost(q, fixed, c)

print(profit(0), profit(20))   # negative, then positive: the answer is between 0 and 20
```

```python
lo, hi = 0.0, 20.0
tol = 1e-8
steps = 0

while hi - lo > tol:
    mid = (lo + hi) / 2
    if profit(mid) < 0:   # still losing money at mid: break-even is to the right
        lo = mid
    else:                 # making money at mid: break-even is to the left
        hi = mid
    steps += 1

q_be = (lo + hi) / 2
print(f"Break-even at q = {q_be:.6f} after {steps} halvings")
```

Check against the quadratic formula: $0.5q^2 - 40q + 200 = 0$ gives $q = 40 - \sqrt{1200} \approx 5.358984$. Each halving cuts the bracket in two, so 31 steps shrink a width of 20 below $10^{-8}$.

The `if` above works because profit rises on this bracket. A general version compares **signs** instead, so it works whichever way the function slopes. Functions can take other functions as inputs:

```python
def bisect(f, lo, hi, tol=1e-8):
    """Return x in [lo, hi] with f(x) close to 0. f(lo) and f(hi) must differ in sign."""
    if f(lo) * f(hi) > 0:
        raise ValueError("f(lo) and f(hi) must have opposite signs")
    while hi - lo > tol:
        mid = (lo + hi) / 2
        if f(lo) * f(mid) <= 0:   # sign change in the left half
            hi = mid
        else:                     # sign change in the right half
            lo = mid
    return (lo + hi) / 2

print(bisect(profit, 0, 20))     # first break-even point
print(bisect(profit, 20, 100))   # profit turns negative again at high output
```

In Week 6 you'll use `scipy.optimize.brentq`, which does the same job in fewer steps. Now you know what's inside it. PS1 asks you to adapt this idea to find a market equilibrium.

## 7. Debugging

Errors are normal; reading them is a skill. When a cell fails, Python prints a **traceback**:

1. **Read the last line first.** It names the error type and gives a message.
2. **Find the arrow** (`---->`). It points to the line that failed.
3. Fix that line, or the line that gave it a bad value, and rerun.

Each cell below fails on purpose. If you use Run All, it may stop at the first one; run the rest of the notebook a cell at a time.

```python
#| error: true
revnue(25, 100, 2)
```

`NameError`: Python doesn't know the name `revnue`. Usually a typo, or a cell defining the name hasn't been run yet (run cells from the top).

```python
#| error: true
revenue(25, 100)
```

`TypeError`: the function needs three arguments and got two.

| Error | Usual cause |
|---|---|
| `NameError` | typo, or the defining cell wasn't run |
| `TypeError` | wrong number or kind of arguments, e.g. adding text to a number |
| `SyntaxError`, `IndentationError` | missing colon, bracket, or quote; inconsistent indentation |
| `ZeroDivisionError` | dividing by zero, often at a corner of a model, such as zero quantity |
| `IndexError`, `KeyError` | asking a list for a position it doesn't have, or a dictionary for a missing name |

### Bugs that don't raise errors

The most dangerous bugs produce a number that looks fine. This demand function has the wrong sign and runs without complaint:

```python
def demand_buggy(p, a, b):
    """Quantity demanded at price p."""
    return a + b * p    # should be a - b*p

demand_buggy(10, 100, 2)
```

The defense is to test a case you can do in your head. An `assert` stops with an error if its condition is false, so a wrong function can't slip by. PS self-check cells work this way.

```python
#| error: true
assert demand_buggy(0, 100, 2) == 100              # passes: at P = 0 the sign doesn't matter
assert demand_buggy(10, 100, 2) == 80, "demand should fall as price rises"
```
